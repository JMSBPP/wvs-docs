I am writing the replicating portfolio for my buyer persona. He starts with initial wealth in cash $W^{\$}_0$, has access to the money market at a borrowing interest rate $i_{\$} \in (0,1)$, and has the desired payoff:

$$
\Pi = 
\begin{cases}
    0 & P < \psi_{\text{ask}} \\
    \\
    P_1 (H) - \psi_{\text{ask}} & P \geq \psi_{\text{ask}}
\end{cases}
$$


 The current spot price is $P_0 < \psi_{\text{ask}}$. Assuming no slippage, $\Delta_0$ denotes the units of the underlying that the trader shorts or longs at $t=0$to achieve the desired payoff (to be determined).

This is:

$$
\begin{cases}
    W_0^{\$} - P_0\cdot \Delta_0 > 0 && \text{borrow/longing}
    \\
    W_0^{\$} - P_0\cdot \Delta_0 > 0 && \text{lending/shorting}
\end{cases}
$$
All at the interest rate $i_{\$}$

Now $P_1 (T) < P_0 < P_1(H)$, then
$$
\begin{cases}
    W_1 (T) = \Delta_0\cdot P_1 (T) + (1+i_{\$})(W_0^{\$} - \Delta_0\cdot P_0) \\
    \\
    W_1 (H) = \Delta_0\cdot P_1 (H) + (1+i_{\$})(W_0^{\$} - \Delta_0\cdot P_0)
\end{cases}
$$

Ok then, to achieve it's payoff $\Pi$, the trader needs to determine $\Delta_0$ and $W_0$ such that:

$$

W_1 (T) = \Pi(P < P_0) = 0 \\
\\

W_1 (H) = \Delta_0 \cdot \Pi(P < P_0) = \Delta_0\cdot P_1 (H) - \psi_{\text{ask}} \big (\big)
$$
