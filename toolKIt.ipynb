We have some minimal restrictions on the admissible pair

$$
\mathcal{A} (x_{ij}, x_{ik}) \subseteq \mathbb{R}^2_{>0}
$$


Formally this set is:

$$
\begin{align*}
\mathcal{A} (x_{ij}, x_{ik}) &= \bigg [0, \frac{\overline{M}}{P_{ij}^{\$}} \bigg] \times \bigg [0, x^{*}_{ik}\bigg ] \\
&= 
\bigg \{ \big (\Delta_{x_{ij}}, \Delta_{x_{ik}}\big) : 0 \leq \Delta_{x_{ij}} \leq \frac{\overline{M}}{P_{ij}^{\$}}, \, 0 \leq \Delta_{x_{ik}} \leq x^{*}_{ik} \bigg \}
\end{align*}
$$
Now we introduce trajectories over $\mathcal{A} (x_{ij}, x_{jk})$ 

Then we have:

$$
\begin{align*}
    \mathcal{v}: \mathcal{A} \to \mathcal{A} \\
    \mathcal{v}_{x_{ij}} : \big [ 0, \frac{\overline{M}}{P_{ij}^{\$}} \big ] \to \mathcal{A} \\
    \mathcal{v}_{x_{ik}} : \big [ 0, x^*_{ik}\big ] \to \mathcal{A}
\end{align*}
$$

Now, we want to introduce more restrictions on $\mathcal{A}$ using the tools enabled by $\mathcal{v}$. The first restriction is 

$$
\begin{align*}
    - \Delta \mathcal{v}_{x_{ij}} = |p|\cdot \Delta \mathcal{v}_{x_{ik}} , \, |p|\in \mathbb{R} 
\end{align*}
$$
