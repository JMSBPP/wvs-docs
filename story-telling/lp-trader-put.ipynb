# LP Sales Pitch: AMMs as Continuum of Put Options

**LP:** 

- I am selling "merchant" price insurance for selling $R_X$ regardles of $P_T$.

I need $R_X$ units of $X$ at $\psi_{\text{bid}}$, no more. I am deeply convinced that is a fair price for $X$. Anything above is too expensive.

If I can get $R_X$ paying $P \leq \psi_{\text{bid}}$ I am good. Well the simplest is let's set a limit order on a exchange to buy $R_X$ at $P \leq \psi_{\text{bid}}$.

However timing the market is not such a good practice. After all I can be wrong and my order never get executed. Is there a way to set this order without timing the market. How do I trade my belief ?





- I am buying $R_X$ at $\psi_{\text{bid}}$. Then if $P_T < \psi_{\text{bid}}$ you a sure buyer at $\psi_{\text{bid}}$ earning a realized premia of $\psi_{\text{bid}} - P_T > 0 $.

- If $P_T > \psi_{\text{bid}}$ do not worry you sell your $\Delta X \leq R_X$ at $P_T$, and this never gets executed
$$
\begin{cases}
    \mathsf{buyer} := \left(\psi_{\text{bid}} - P_T\right)^+ - \mathsf{premia} \\
    \mathsf{seller} := \mathsf{premia} - \left(\psi_{\text{bid}} - P_T\right)^+
\end{cases}
$$
